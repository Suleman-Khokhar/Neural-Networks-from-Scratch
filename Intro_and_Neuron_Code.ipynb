##<h2>Parameters in Neural Networks <h2>

<b> Parameters are the values inside a neural network that the network learns and changes during training to improve its predictions. <b>

The two main parameters are:

Weights <br>
Biases
1. Weights

A weight tells the neuron how important an input is.

If a weight is large, that input has a stronger effect on the neuron's output.

2. Bias

A bias is an additional value added to the weighted inputs.

The basic neuron formula is:

z= x1*w1 + x2*w2 +b

For example:

x1 = 2
w1 = 0.5

x2 = 3
w2 = 0.8

b = 1

Then:

z=(2*0.5)+(3*0.8)+1  
z=4.4

Then we pass z through an activation function.

<b>Formula of parameter <b> <br>
   p = w + b <br>

<h4>General form : <h4>parameter = (input_of_1layer * number_of_neurons_in_next_layer + number_of_biases_in_next_layer) + (input_of_2layer * number_of_neurons_in_next_layer + number_of_biases_in_next_layer) + ...... up to end  <br>

where w = total weights and b = total biases.  This is the standard definition — a "parameter" in a neural network is anything the network learns during training, and that's exactly weights + biases.

## Class implementation


````python
import numpy as np

np.random.seed(0)

X = [[1, 2, 3, 2.5],
     [2.0, 5.0, -1.0, 2.0],
     [-1.5, 2.7, 3.3, -0.8]]


class Layer_Dense:
    def __init__(self, n_inputs, n_neurons):
        self.weights = 0.10 * np.random.randn(n_inputs, n_neurons)
        self.biases = np.zeros((1, n_neurons))
    def forward(self, inputs):
        self.output = np.dot(inputs, self.weights) + self.biases

layer1 = Layer_Dense(4,5)
layer2 = Layer_Dense(5,2)

layer1.forward(X)
#print(layer1.output)
layer2.forward(layer1.output)
print(layer2.output)
````

````python
import numpy as np

# Set a pseudo-random seed so that the random numbers generated
# are identical every time you run this program.
np.random.seed(0)

# Input dataset (Batch of 3 samples, where each sample has 4 features).
# Matrix shape: (3, 4) -> (samples, features)
X = [[1, 2, 3, 2.5],
     [2.0, 5.0, -1.0, 2.0],
     [-1.5, 2.7, 3.3, -0.8]]

# Define the object-oriented structure for a fully connected (Dense) layer.
class Layer_Dense:
    # Initializer method executed upon creating a layer instance.
    def __init__(self, n_inputs, n_neurons):
        # Initialize weights with random values from a Gaussian distribution.
        # Shape: (n_inputs, n_neurons)
        # We multiply by 0.10 to keep initial weights small, preventing values
        # from exploding as they move deep into the network.
        # Defining it this way completely removes the need for a manual .T transpose!
        self.weights = 0.10 * np.random.randn(n_inputs, n_neurons)

        # Initialize biases as a horizontal row vector filled with zeros.
        # Shape: (1, n_neurons) -> One unique bias value per neuron.
        self.biases = np.zeros((1, n_neurons))

    # Forward pass method to process input data through this layer.
    def forward(self, inputs):
        # Calculate the neuron equation: Output = (Inputs . Weights) + Biases
        # Inputs shape:  (3, 4)
        # Weights shape: (4, 5)  <- Inner dimensions (4 and 4) align perfectly!
        # Dot product result: (3, 5)
        # Adding biases row vector (1, 5) broadcasts over all 3 matrix rows.
        self.output = np.dot(inputs, self.weights) + self.biases

# Create the first hidden layer.
# It takes 4 inputs (matching features in X) and contains 5 neurons.
# Internal weights shape: (4, 5)
layer1 = Layer_Dense(4, 5)

# Create the second hidden layer.
# It must take 5 inputs because layer1 outputs data from 5 neurons.
# It contains 2 neurons (the final output parameters).
# Internal weights shape: (5, 2)
layer2 = Layer_Dense(5, 2)

# Pass the raw input data 'X' forward through layer 1.
# This calculates and populates 'layer1.output' with a shape of (3, 5).
layer1.forward(X)

# Cascade the data forward by feeding layer1's output into layer 2.
# This calculates and populates 'layer2.output' with a shape of (3, 2).
layer2.forward(layer1.output)

# Print out the final predictions/evaluations matrix from layer 2.
# Expected shape: (3 samples, 2 neuron outputs)
print(layer2.output)
````

## Activation Function (Adding Non-Linearity):
The linear output is passed through an activation function (like ReLU, which turns all negative values to zero) <br>
Without this, the network would just be a giant linear equation and couldn't learn complex patterns like curves or images.

Passing to the Next Layer:
<br> The activated output becomes the input for the next layer. This repeats until the final output layer generates raw predictions.

## Phase
 <br>
The Bridge (Calculating Loss) <br>
Before going backward, the network needs to score its performance. <br>
• The final layer's predictions are compared against the true targets (ground truth) using a Loss Function (like Mean Squared Error or Categorical Cross-Entropy). <br>
• The loss function outputs a single number: The Error Score. A high loss means bad predictions; a low loss means great predictions.

### This is the core flow of how a neural network learns. You already understand weights, biases, shapes, and dot products, so now we can connect them into one complete process.

The easiest way to remember it is:

<br>
Forward pass = make a prediction.<br>
Loss = measure how wrong the prediction is.<br>
Backward pass = figure out how each parameter contributed to the error.<br>
Update = change the parameters to reduce the error.<br>

Then repeat.

## Step 1: The Perceptron (single neuron)

A perceptron takes inputs, weights them, sums them, and passes the result through an activation function.

Linear part (the "z"):


$$z = w_1x_1 + w_2x_2 + \dots + w_nx_n + b = \mathbf{w}^\top\mathbf{x} + b$$
$w_i$ = weight (how important input $x_i$ is)
$b$ = bias (shifts the decision boundary)
This is just a dot product + offset — geometrically, a line (2D) or hyperplane (higher-D)

Worked example from your slides: $w_0=1, w_1=3, w_2=-2$


$$\hat{y} = g(1 + 3x_1 - 2x_2)$$

At $x=[-1,2]$: $z = 1+3(-1)-2(2) = -6$


## Step 2: Activation function (the "g" or "σ")

Without this, stacking layers is pointless — a linear function of a linear function is still linear. Non-linearity is what lets a network carve curved decision boundaries instead of just a straight line.

Sigmoid: $g(z)=\frac{1}{1+e^{-z}}$ → squashes to (0,1), used for probabilities, 2-class output
Tanh: $g(z)=\frac{e^z-e^{-z}}{e^z+e^{-z}}$ → squashes to (-1,1)
ReLU: $g(z)=\max(0,z)$ → cheap, avoids vanishing gradient, default for hidden layers
Softmax: generalizes sigmoid to k>2 classes, outputs a probability vector summing to 1

Continuing the example: $g(-6) = \sigma(-6) \approx 0.002$ → network predicts class 0.

Where Sigmoid Is Used

1. Output layer for binary classification — squashes the raw score $z$ into a probability between 0 and 1, which you interpret as $P(y=1|x)$. This is exactly what your Week 1 slides called the "classification function."

2. Inside logistic regression — it's the $h_\theta(x)$ in everything we derived yesterday (cross-entropy loss, gradient descent, backprop all assumed sigmoid).

3. Gates in LSTMs (later in your course) — sigmoid decides "how much information to let through" (0 = block, 1 = pass), since its output naturally acts like a switch/percentage.

4. Historically, as a hidden-layer activation — mostly replaced by ReLU now because sigmoid saturates and kills gradients in deep networks (vanishing gradient problem).

Worked Example

Take your Week 2 perceptron: $z = 1 + 3x_1 - 2x_2$, input $x = [-1, 2]$:


$$z = 1 + 3(-1) - 2(2) = 1 - 3 - 4 = -6$$
$$\sigma(-6) = \frac{1}{1+e^{6}} \approx \frac{1}{1+403.4} \approx 0.00248$$

So the model outputs $\approx 0.25\%$ probability of class 1 — confidently predicting class 0. Notice: a large negative $z$ gets squashed near 0, a large positive $z$ gets squashed near 1, and $z=0$ gives exactly 0.5 (the decision boundary).
